# Recursion
Function calls itself

## Advantages:
No loops needed. <br>
Solves problems without iteration. <br>
*"To understand recursion you must understand recursion"*

## Iterative Vs Recursive

```python
# iterative method
def multiply(a, b):
    result = 0
    for i in range(b):
        result += a
    print(result)
multiply(3, 4)
```

```python
# Recursive Method

# 1. Base Case: Define stopping condition.

# 2. Decompose: Break main problem into smaller subproblems until base case is reached.

def mul(a, b):
    if b == 1:
        return a
    else:
        return a + mul(a, b-1)
```

```python
print(mul(3, 4))
```

```python
# Factorial via Recursion
def fact(number):
    if number == 1:
        return 1
    else:
        return number * fact(number-1)
print(fact(5))
```

```python
# Palindrome
def palin(text):
    if len(text) <= 1:
        print("palindrome")
    else:
        if text[0] == text[-1]:
            palin(text[1:-1])
        else:
            print("not a palindrome")
```

```python
palin("madam")
```

```python
palin("python")
```

# Memoization
Memoization refers to remembering method call results based on inputs. <br> <br>

Returns cached results, avoiding recomputation. <br>
Speeds up computations; stores previous results. <br>
Used in dynamic programming for recursive solutions. <br>
Reduces time complexity; avoids redundant calculations. <br>
Optimizes recursive algorithms by reusing results

```python
def memo(m, d):
    if m in d:
        return d[m]
    else:
        d[m] = memo(m-1, d) + memo(m-2, d)
        return d[m]
d = {0:1, 1:1}
print(memo(48, d))
```
