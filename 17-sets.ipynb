## Frozen set
Immutable set. <br>
Created via frozenset(). <br>
No modifications post-creation. <br>
Ideal for hashable items: **strings, numbers**.

```python
# create frozenset
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4, 5])
fs1 | fs2
```

```python
# what works and what does not
# Works       ---> all read functions
# Doesn't Work ---> write operations
```

```python
# When to use
# 2D sets
fs = frozenset([1, 2, frozenset([3, 4])])
fs
```
