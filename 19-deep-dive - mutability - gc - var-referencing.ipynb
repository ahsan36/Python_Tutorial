# Garbage Collection
**Unused memory** remains after references are deleted. <br>

**Garbage Collector** manages and frees this space. <br>

Periodic deletion of unreferenced values is called **Garbage Collection** <br>

# Wierd Stuff
### *Python's Wierd Behaviour/Oddities:*
**1.Ref Count Anomaly/Getrefcount Anomaly:** sys.getrefcount() shows an extra count due to temp ref. <br>
**2. Integer Caching:** Caches -5 to 256 integers for optimization. <br>
**3.String Interning:** Same string literals share memory.

## Mutability ---> Ability to modify data at its memory location.
Depends on Data type.

### Immutable Data Types
str, int, float, bool, complex, tuple 
### Mutable Data Types
list, dict, set <br> <br>

**Immutable Data Types:** <br>
Ops = new objs, new mem locs, new refs. <br> <br>

**Mutable Data Types:** <br>
Ops = in-place, same addr.

## Cloning
In cloning we create a copy of the list at a different memory address. <br> <br>
append, edit, insert, extend modifies list in place (mutable); same memory address.

Concatenation creates new list; different memory address.
