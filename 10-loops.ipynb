# Loops in Python
Repeat code block. <br>
*Iteration:* execute same code repeatedly.<br>

**While Loop** <br>
Executes while condition true.

**For Loop** <br>
Executes for each item in iterable.

### While Loop

```python
# Syntax
while condition:
    code
```

```python
# Print multiplication table

number = int(input("Entr the number: "))

i = 1
while i < 11:
    print(number * i)
    i += 1
```

```python

number = int(input("Enter the number: "))
i = 1
while i < 11:
    print(number, "*", i, "=", number * i)
    i += 1
```

```python
# while loop with else

x = 1
while x < 3:
    print(x)
    x += 1
else:
    print("Limit crossed")
```

```python
# 2. Sum of Digits
number = int(input("Enter the number: "))

sum_digits = 0

while number > 0:
    digit = number %  10
    sum_digits += digit
    number //= 10
print(f"Sum of digits: {sum_digits}")
```

### For Loop

```python
for item in iterable:
    # Code block
    
# iterable: list, tuple, string, etc.
```

```python
# Range function
range(1, 11)
```

```python
list(range(1, 11))    # 1 to 10
```

```python
list(range(5))    # 0 to 4
```

```python
list(range(1, 11, 2))    # 0 to 10 (difference of 2)
```

```python
list(range(1, 11, 3))    # 0 to 10 (difference of 3)
```

```python
list(range(10, 0, -1))   # 10 to 1
```

```python
for i in range(1, 11):
    print(i)
```

```python
for i in range(1, 11, 2):
    print(i)
```

```python
for i in range(10, 0, -1):
    print(i)
```

```python
for i in "India":
    print(i)
```

```python
for i in [1, 2, 3, 5]:
    print(i)
```

```python
for i in (1, 2, 3, 5):
    print(i)
```

```python
for i in {1, 2, 3, 5}:
    print(i)
```

```python
# 2. Sequence Sum

# Formula: 1/1! + 2/2! + 3/3! + ...

n = int(input('enter n: '))

result = 0;
fact = 1;

for i in range(1, n+1):
    fact *= i
    result += i / fact
print(result)
```

**Key Differences** is **For Loop** iterates a fixed number of times; **While Loop** iterates until condition is False.

**For Loop Usage** - Known iterations/sequence traversal; **While Loop Usage** - Uncertain iterations/condition-based.

### Nested Loops

```python
# Uniqe Pairs Example

for i in range(1, 5):
    for j in range(1, 5):
        print(i, j)
```

```python
# *
# * *
# * * *
# * * * *
# * * * * *

rows = int(input("Enter rows: "))

for i in range(1, rows+1):
    for j in range(0, i):
        print("*", end=" ")
    print("")
```

```python
# 1
# 121
# 12321
# 1234321

rows = int(input('Enter rows: '))
for i in range(1, rows+1):
  for j in range(1, i+1):
    print(j, end='')
  for k in range(i-1, 0, -1):
    print(k, end='')
  print()
```
