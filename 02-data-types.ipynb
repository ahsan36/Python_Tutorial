# Python Data Types categories

## Basic Types

**.Int:** Whole num(1,2,3) <br>
**.Float:** Decimals (3.14, 2.5) <br>
**.Complex:** Real+Imag (1 + 2j) <br>
**.Bool:** True/False <br>
**.Str:** Text ("hello") <br>

## Container Types:

**.List :** Order items. <br>
[1, 2, 3] <br>
**.Tuple :** Ordered, immutable.<br>
(1, 2, 3) <br>
**.Set :** Unordered, unique. <br>
{1, 2, 3} <br>
**.Dict :** Key-value pairs. <br>
{'k1': 'v1', 'k2': 'v2'}
