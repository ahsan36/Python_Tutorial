# Operators in Python 
**Arithmetic:** +, -, *, /, %, **, // <br>
**Comparison:** ==, !=, >, <, >=, <= <br>
**Logical:** and, or, not <br>
**Bitwise:** &, |, ^, ~, <<, >> <br>
**Assignment:** =, +=, -=, *=, /=, %=, **=, //= <br>
**Identity:** is, is not <br>
**Membership:** in, not in

```python
print(x * y)
```

```python
print(x / y)
```

```python
print(x % y)
```

## Bitwise Operators

```python
x = 2    # 010
y = 3    # 110
```

```python
print(x & y) # AND
```

```python
print(x | y) # OR
```

```python
print(2 ^ 3) # XOR
```

```python
print(x >> 2) # Right Shift
```

```python
print(y << 3) # Left Shift
```

```python
print(~ x) # NOT
```

## Assignment Operators

```python
a = 3
print(a)
```

```python
a += 3   # a = a + 3
print(a)
```

```python
a -= 3   # a = a - 3
print(a)
```

```python
a *= 3   # a = a * 3
print(a)
```

```python
a /= 3   # a = a / 3
print(a)
```

```python
# Invalid syntax

a++
++a
```

```python
a = a + 3
a += 3
```

## Identity Operators

```python
a = 3
b = 3
print(a is b)
```

```python
a = "Hello"
b = "Hello"
print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)
```

```python
# Above False bcz not in the same memory location
```

```python
a = "Hello-World"
b = "Hello-World"
print(a is b)
print(a is not b)
```

## Membership Operators

```python
x = "Delhi"
print("D" in x)
print("D" not in x)
```

```python
x = [1, 2, 3]
print(1 in x)
print(5 in x)
```

```python
x = (1, 2, 3)
print(1 in x)
print(5 in x)
```
