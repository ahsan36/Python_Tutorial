# Decision Control
Control the flow of program execution based on specific conditions.

## if-else

```python
age = int(input("Enter your age: "))

if age < 18:
    print("Not Adult")
elif age >= 18 and age <=50:
    print("Adult")
else:
    print("old")
```

```python
# correct email - ahsan@gmail.com
# password = 1234

email = input("Enter your email: ")
if "@" in email:
    password = input("Enter your password: ")

    if email == "ahsan@gmail.com" and password == "1234":
        print("Welcome")
    elif email == "ahsan@gmail.com" and password != "1234":
        print("Password incorrect")
        password = input("Enter your password again: ")
        if password == "1234":
            print("Finally correct")
        else:
            print("Still incorrect")
    else:
        print("Incorrect credentials")
else:
    print("Email is incorrect, Please enter correct email: ")
```
