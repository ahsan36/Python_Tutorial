Can not use as **variable names**.

```python
# Python has 33 keywords.
import keyword
print(keyword.kwlist)
```

## Identifiers
**Identifiers** are name for var, func, class etc.

**Rules:** <br>
Start with letter or _ (underscore) <br>
Followed by letter, digit, _ <br>
Not a python keyword <br>

```python
name = "Ahsan"
print(name)
```

```python
_ = "Ahsan"
print(_)
```

```python
first_name = "Ahsan"
print(first_name)
```

```python
False = "Ahsan" # error show bcz False is a python keyword, we can not name as a variable
print(False)
```
