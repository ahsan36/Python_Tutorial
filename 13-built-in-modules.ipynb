# What are Modules?
 **Code libraries** in Python. <br>
 **Contain functions** for use. <br><br>
*Examples:* <br><br>
  **Math:** Math ops <br>
  **Random:** Random numbers <br>
  **os:** OS interaction <br>
  **time:** Time tasks

```python
help('modules')
```

```python
# keyword module
import keyword
print(keyword.kwlist)  # Reserved keywords in Python
```

```python
# Math Module
import math
```

```python
math.pi
```

```python
math.e
```

```python
math.factorial(5)
```

```python
math.ceil(6.3)
```

```python
math.floor(6.9)
```

```python
# Random Module
import random
```

```python
random.randint(1, 100)
```

```python
a = [1, 2, 3, 4, 5]
random.shuffle(a)
a
```

```python
# Time Module
import time
```

```python
time.time() # Current time (seconds)
```

```python
time.ctime() # Current time (string)
```

```python
print("Hello")
time.sleep(5) # Delay for 5 seconds
print("World")
```

```python
# Datetime Module

import datetime
print(datetime.datetime.now()) # Current date/time
```

```python
# OS Module
import os
```

```python
os.getcwd() # Current working dir
```

```python
os.listdir() # Files in dir
```
