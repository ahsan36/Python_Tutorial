## List vs Array
**Array**: Homogeneous; Contiguous memory; Fast; Numerical/Scientific use.<br>
**List**: Heterogeneous; Non-contiguous memory; Programmer-friendly; General-purpose.<br><br>

1.Create <br>
2.Access <br>
3.Edit <br>
4.Add <br>
5.Delete <br>
6.Operations <br>
7.Functions <br>
## Arrays vs. Lists in Memory:
**Arrays:** <br>

int arr[50]; ---> 50 contiguous memory blocks. <br>
Elements stored in binary form at consecutive addresses. <br>
**Lists:** <br>

list_example = [1, 2, 3] ---> Elements at different locations. <br>
Stores references/pointers to elements, not the values. <br>

### List Comprehension
Compact list creation.<br>

newlist = [expresion **for** item in iterable **if** condition == **True**]<br>
Example:<br>

squares = [x**2 for x in range(10)] ---> squares 0-9.<br>

**Advantages**<br>

Efficient: Time & space.<br>
Concise: Fewer lines.<br>
Formulaic: Iteration ---> expression.

```python
L = [1, 2, 3, 4, 5, 6, 7]
L
```

```python
L1 = [item * 2 for item in L]
L1
```

```python
L2 = [i**2 for i in range(10)]
L2
```

```python
L3 = [i**2 for i in range(10) if i%2!=0]
L3
```

```python
fruits = ['Apple', 'Orange', 'Mango', 'Guava']
fruits
```

```python
L4 = [fruit for fruit in fruits if fruit[0]=="O"]
L4
```

```python
# Add 1-10 to list
L = []
for i in range(1, 11):
  L.append(i)
print(L)
```

```python
L = [i for i in range(1, 11)]
print(L)
```

```python
# Scalar Multiplication
v = [2, 3, 4] # Vector
s = -3        # Scalar
# [-6, -9, -12]
[s*i for i in v]
```

```python
# Add squares
L = [1, 2, 3, 4, 5]
[i**2 for i in L]
```

```python
# Print nums divisible by 5 from 1 to 50
[i for i in range(1, 51) if i%5 == 0]
```

```python
# Nested If with List Comprehension
basket = ['apple', 'guava', 'cherry', 'banana']
my_fruits = ['apple', 'kiwi', 'grapes', 'banana']
# Add fruits from `my_fruits` that are in `basket` and start with 'a'
[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
```

```python
# 3x3 Matrix via Nested List Comprehension
[[i*j for i in range(1, 4)] for j in range(1, 4)]
```

```python
# Cartesian Products ---> List Comprehension on 2 lists together
L1 = [1, 2, 3, 4]
L2 = [5, 6, 7, 8]
[i*j for i in L1 for j in L2]
```

### List Traversal
Itemwise<br>
Indexwise

```python
# Itemwise
L = [1, 2, 3, 4]
for i in L:
  print(i)
```

```python
# Indexwise
L = [1, 2, 3, 4]
for i in range(0, len(L)):
  print(L[i])
```

## Zip
**zip() Function:** <br>

Yields zip obj: iterator of tuples.<br>
Combines 1st items of each iterator, then 2nd, etc.<br>
Example:<br>
list1 = [1, 2, 3]<br>
list2 = ['a', 'b', 'c']<br>
zipped = zip(list1, list2)<br>
print(list(zipped))  *# Output: [(1, 'a'), (2, 'b'), (3, 'c')]* <br><br>
**Length Mismatch:** <br>

Shortest iterator determines output length.

```python
# Add items of 2 lists indexwise
L1 = [1, 2, 3, 4]
L2 = [-1, -2, -3, -4]
list(zip(L1, L2)) 
[i+j for i, j in zip(L1, L2)]
```
