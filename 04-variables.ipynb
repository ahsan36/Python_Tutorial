## Variables
Name of the memory location that is use to hold/store data.

### Dynamic Typing
Don't need to specify data type, while declaring a variable. <br>
Python will automatically reduce.

```python
name = "Ahsan"
print(name)
```

### Dynamic Binding
Type associated at runtime.
Changeable during execution.

```python
a = 4       # Assign integer

print(a)

a = "Ahsan" # Reassign string

print(a)
```

```python
# Special Syntax

a=3; b=4; c=5

print(a)
print(b)
print(c)
```

```python
a, b, c = 3, 4, 5

print(a)
print(b)
print(c)
```

```python
a = b = c = 5

print(a)
print(b)
print(c)
```
