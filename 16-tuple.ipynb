Immutable list-like collection.

Ordered elements. <br>
Unmodifiable post-creation.<br> 
Allows duplicates. <br> <br>

## Lists vs Tuples
### Syntax:
Lists: [ ] <br>
Tuples: ( ) <br>
### Mutability:
Lists: Mutable <br>
Tuples: Immutable <br>
### Speed:
Lists: Slower (mutable) <br>
Tuples: Faster (immutable) <br>
### Memory:
Lists: Higher <br>
Tuples: Lower <br>
### Functionality:
Both: Indexing, slicing <br>
Lists: More methods <br>
### Error-Prone:
Lists: Modifiable <br>
Tuples: Safer <br>
### Use Case:
Lists: Dynamic <br>
Tuples: Static

## Q. Why use Tuples?
Immutable; prevents changes. <br>
Ensures data integrity. <br>
Use for fixed collections. <br>
Example: <br>
*college_database = ('CS', 'Math', 'Physics')* <br>
 college_database[0] = 'Electronics'  # TypeError <br>
Use for static data; lists for mutable.

```python
a = [1, 2, 3]
b = a
a.append(4)
print(a)
print(b)
```

```python
a = (1, 2, 3)
b = a
a = a + (4,)
print(a)
print(b)
```

```python
# tuple unpacking
a, b, c = (1, 2, 3)
print(a, b, c)
```

```python
a, b = (1, 2, 3)
print(a, b)
```

```python
a = 1
b = 2
a, b = b, a
print(a, b)
```

```python
a, b, *others = (1, 2, 3, 4)
print(a, b)
print(others)
```

```python
# zipping tuples
a = (1, 2, 3, 4)
b = (5, 6, 7, 8)
tuple(zip(a, b))
```
