# Type Conversion

**Implicit :** Auto type conversion by language. <br>
**Explicit :** Manual type conversion usinf functions/methods.

```python
# Implicit Conversion

print(4 + 3.6)   # Implicit: int + float -> float

print(type(4), type(3.6))  # SHow type: int, float

print(3 + '3')  # Error: no implicit int + str
```

```python
# Explicit Conversion

# str to int
int('5')
```

```python
# int to str
str(4)
```

```python
# float
float(6)
```
