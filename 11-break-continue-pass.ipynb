## Break, Continue, Pass
**Break:** End loop immediately <br>
**Continue:** Skips to next iteration <br>
**Pass:** Placeholder; does nothing <br>

```python
# Break

for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

```python
# Continue

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
# Pass

for i in range(1, 11):
    pass
```
